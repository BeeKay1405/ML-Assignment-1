```python
import pandas as pd
```

```python
trainVar1 = pd.read_csv("BT2024269_train_var1.csv")
testVar1 = pd.read_csv("BT2024269_test_var1.csv")

trainVar2 = pd.read_csv("BT2024269_train_var2.csv")
testVar2 = pd.read_csv("BT2024269_test_var2.csv")
```

```python
print("TRAIN VAR1")
print(trainVar1.shape)
print(trainVar1.columns)
print(trainVar1.head())

print("\nTEST VAR1")
print(testVar1.shape)
print(testVar1.columns)
print(testVar1.head())

print("\nTRAIN VAR2")
print(trainVar2.shape)
print(trainVar2.columns)
print(trainVar2.head())

print("\nTEST VAR2")
print(testVar2.shape)
print(testVar2.columns)
print(testVar2.head())
```

```python
print("VAR1 TRAIN INFO")
print(trainVar1.info())

print("\nVAR2 TRAIN INFO")
print(trainVar2.info())

print("\nVAR1 MISSING VALUES")
print(trainVar1.isnull().sum())

print("\nVAR2 MISSING VALUES")
print(trainVar2.isnull().sum())

print("\nVAR1 STATISTICS")
print(trainVar1.describe())

print("\nVAR2 STATISTICS")
print(trainVar2.describe())
```

```python
import matplotlib.pyplot as plt

print("VAR1 CORRELATION MATRIX")
display(trainVar1.corr().round(2))

print("\nVAR2 CORRELATION MATRIX")
display(trainVar2.corr().round(2))
```

```python
xVar1 = trainVar1.drop(columns=["y"])
yVar1 = trainVar1["y"]

xVar2 = trainVar2.drop(columns=["y"])
yVar2 = trainVar2["y"]
```

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import Pipeline
```

```python
kFolds = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
model = Pipeline([
    ("polynomial", PolynomialFeatures(degree=1)),
    ("regression", LinearRegression())
])
```

```python
cvResults = cross_validate(
    model,
    xVar1,
    yVar1,
    cv=kFolds,
    scoring=["neg_mean_squared_error", "r2"]
)

print("CV MSE:", -cvResults["test_neg_mean_squared_error"].mean())
print("CV MSE std:", cvResults["test_neg_mean_squared_error"].std())
print("CV R2:", cvResults["test_r2"].mean())
print("CV R2 std:", cvResults["test_r2"].std())
```

```python
degreeResultsVar1 = []

for degree in range(1, 11):
    model = Pipeline([
        ("polynomial", PolynomialFeatures(degree=degree, include_bias=False)),
        ("regression", LinearRegression())
    ])
    
    cvResults = cross_validate(
        model,
        xVar1,
        yVar1,
        cv=kFolds,
        scoring=["neg_mean_squared_error", "r2"]
    )
    
    degreeResultsVar1.append({
        "degree": degree,
        "mseMean": -cvResults["test_neg_mean_squared_error"].mean(),
        "mseStd": cvResults["test_neg_mean_squared_error"].std(),
        "r2Mean": cvResults["test_r2"].mean(),
        "r2Std": cvResults["test_r2"].std()
    })

degreeResultsVar1 = pd.DataFrame(degreeResultsVar1)

degreeResultsVar1
```

```python
scaledResultsVar1 = []

for degree in range(1, 11):
    model = Pipeline([
        ("polynomial", PolynomialFeatures(degree=degree, include_bias=False)),
        ("scaler", StandardScaler()),
        ("regression", LinearRegression())
    ])
    
    cvResults = cross_validate(
        model,
        xVar1,
        yVar1,
        cv=kFolds,
        scoring=["neg_mean_squared_error", "r2"]
    )
    
    scaledResultsVar1.append({
        "degree": degree,
        "mseMean": -cvResults["test_neg_mean_squared_error"].mean(),
        "mseStd": cvResults["test_neg_mean_squared_error"].std(),
        "r2Mean": cvResults["test_r2"].mean(),
        "r2Std": cvResults["test_r2"].std()
    })

scaledResultsVar1 = pd.DataFrame(scaledResultsVar1)

scaledResultsVar1
```

```python
kResultsVar1 = []

for k in [3, 5, 10]:
    kFoldsExperiment = KFold(
        n_splits=k,
        shuffle=True,
        random_state=42
    )
    
    model = Pipeline([
        ("polynomial", PolynomialFeatures(degree=4, include_bias=False)),
        ("regression", LinearRegression())
    ])
    
    cvResults = cross_validate(
        model,
        xVar1,
        yVar1,
        cv=kFoldsExperiment,
        scoring=["neg_mean_squared_error", "r2"]
    )
    
    kResultsVar1.append({
        "k": k,
        "mseMean": -cvResults["test_neg_mean_squared_error"].mean(),
        "mseStd": cvResults["test_neg_mean_squared_error"].std(),
        "r2Mean": cvResults["test_r2"].mean(),
        "r2Std": cvResults["test_r2"].std()
    })

kResultsVar1 = pd.DataFrame(kResultsVar1)

kResultsVar1
```

```python
degreeResultsVar2 = []

for degree in range(1, 21):
    model = Pipeline([
        ("polynomial", PolynomialFeatures(degree=degree, include_bias=False)),
        ("regression", LinearRegression())
    ])
    
    cvResults = cross_validate(
        model,
        xVar2,
        yVar2,
        cv=kFolds,
        scoring=["neg_mean_squared_error", "r2"]
    )
    
    degreeResultsVar2.append({
        "degree": degree,
        "mseMean": -cvResults["test_neg_mean_squared_error"].mean(),
        "mseStd": cvResults["test_neg_mean_squared_error"].std(),
        "r2Mean": cvResults["test_r2"].mean(),
        "r2Std": cvResults["test_r2"].std()
    })

degreeResultsVar2 = pd.DataFrame(degreeResultsVar2)

degreeResultsVar2
```

```python
kResultsVar2 = []

for k in [3, 5, 10]:
    kFoldsExperiment = KFold(
        n_splits=k,
        shuffle=True,
        random_state=42
    )
    
    model = Pipeline([
        ("polynomial", PolynomialFeatures(degree=8, include_bias=False)),
        ("regression", LinearRegression())
    ])
    
    cvResults = cross_validate(
        model,
        xVar2,
        yVar2,
        cv=kFoldsExperiment,
        scoring=["neg_mean_squared_error", "r2"]
    )
    
    kResultsVar2.append({
        "k": k,
        "mseMean": -cvResults["test_neg_mean_squared_error"].mean(),
        "mseStd": cvResults["test_neg_mean_squared_error"].std(),
        "r2Mean": cvResults["test_r2"].mean(),
        "r2Std": cvResults["test_r2"].std()
    })

kResultsVar2 = pd.DataFrame(kResultsVar2)

kResultsVar2
```

```python
finalModelVar1 = Pipeline([
    ("polynomial", PolynomialFeatures(degree=4, include_bias=False)),
    ("regression", LinearRegression())
])

finalModelVar1.fit(xVar1, yVar1)


finalModelVar2 = Pipeline([
    ("polynomial", PolynomialFeatures(degree=8, include_bias=False)),
    ("regression", LinearRegression())
])

finalModelVar2.fit(xVar2, yVar2)

print("Final models trained successfully.")
```

```python
predictionsVar1 = finalModelVar1.predict(testVar1)
predictionsVar2 = finalModelVar2.predict(testVar2)

print("Var1 predictions:", predictionsVar1.shape)
print("Var2 predictions:", predictionsVar2.shape)

print("\nFirst 5 Var1 predictions:")
print(predictionsVar1[:5])

print("\nFirst 5 Var2 predictions:")
print(predictionsVar2[:5])
```

```python
submissionVar1 = pd.DataFrame({
    "y": predictionsVar1
})

submissionVar2 = pd.DataFrame({
    "y": predictionsVar2
})

submissionVar1.to_csv("BT2024269_pred_var_1.csv", index=False)
submissionVar2.to_csv("BT2024269_pred_var_2.csv", index=False)

print("Var1:")
print(submissionVar1.head())

print("\nVar2:")
print(submissionVar2.head())
```

```python
from sklearn.linear_model import Ridge
```

```python
ridgeResultsVar1 = []

ridgeAlphas = [0.001, 0.01, 0.1, 1, 10, 100, 1000]

for degree in range(1, 11):
    for alpha in ridgeAlphas:
        ridgeModel = Pipeline([
            ("polynomial", PolynomialFeatures(degree=degree, include_bias=False)),
            ("scaler", StandardScaler()),
            ("ridge", Ridge(alpha=alpha))
        ])

        scores = cross_validate(
            ridgeModel,
            xVar1,
            yVar1,
            cv=kFolds,
            scoring=["neg_mean_squared_error", "r2"]
        )

        ridgeResultsVar1.append({
            "degree": degree,
            "alpha": alpha,
            "mseMean": -scores["test_neg_mean_squared_error"].mean(),
            "mseStd": scores["test_neg_mean_squared_error"].std(),
            "r2Mean": scores["test_r2"].mean(),
            "r2Std": scores["test_r2"].std()
        })

ridgeResultsVar1 = pd.DataFrame(ridgeResultsVar1)

ridgeResultsVar1.sort_values("mseMean").head(10)
```

```python
bestRidgeVar1 = ridgeResultsVar1.sort_values("mseMean").iloc[0]

print("Best Ridge model:")
print("Degree:", int(bestRidgeVar1["degree"]))
print("Alpha:", bestRidgeVar1["alpha"])
print("CV MSE:", bestRidgeVar1["mseMean"])
print("CV MSE std:", bestRidgeVar1["mseStd"])
print("CV R2:", bestRidgeVar1["r2Mean"])
print("CV R2 std:", bestRidgeVar1["r2Std"])
```
